# Anotações Aula 23 - Modelos de Tempo Discreto: ARX e ARMAX

- AR -> Autoregressivo
- ARX -> Autoregressivo com entrada exógena
- Car -> Controlled Auto Regressive
- ARMAX -> Autoregressivo com entrada exógena e média móvel
- ARIMAX -> Adiciona um integrado ao sistema.

ARX e ARMAX serão utilizados na disciplina.

 - Estudar os conceitos apresentados nos slides. As aplicações, porque existem e a diferença da abordagem anterior (se há relação ou não).


```python
import matplotlib.pyplot as plt
import numpy as np


def simulate_arx(t: np.ndarray, u: np.ndarray, e: np.ndarray, a: float, b: float, c: float=0) -> np.ndarray:
    n = len(t)
    y = np.zeros(n)

    for k in range(1, n):
        y[k] = a * y[k - 1] + b * u[k - 1]   + c*e[k]

    return y


def plot_arx(t: np.ndarray, u: np.ndarray, e: np.ndarray, a: float, b: float, c: float=0) -> None:

    y = simulate_arx(t, u, e, a, b, c)

    fig, ax = plt.subplots(figsize=(10, 4))

    ax.plot(t, y, label="y[t]", color="tab:green")
    ax.set_ylabel("Output y[t]")
    ax.set_xlabel("Time")
    ax.legend()
    ax.grid(True)

    fig.suptitle(f"ARX Model: y[t] = {a}·y[t-1] + {b}·u[t-1] {f'+ {c}·e[t]' if c > 0 else ''}")
    plt.tight_layout()
    plt.show()

```

```python
a = 0.85
b = 0.15
c = 0


if __name__ == "__main__":
    N = 30
    t = np.arange(N)
    u = np.ones(N)
    # u[:50] = 0
    e = np.random.normal(0, 0.1, N)

    plot_arx(t, u, e, a, b, c)
```
